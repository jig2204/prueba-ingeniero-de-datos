```python
from pathlib import Path

def obtener_html(entradas):
    archivos_html = []

    for entrada in entradas:
        ruta = Path(entrada)


        if ruta.is_file() and ruta.suffix.lower() == ".html":
            archivos_html.append(str(ruta))

        elif ruta.is_dir():
            for archivo in ruta.rglob("*.html"):
                archivos_html.append(str(archivo))

    return archivos_html

entradas= ["archivos_htlm"]

html_encontrados = obtener_html(entradas)

for archivo in html_encontrados:
    print(archivo)
```

```python
from pathlib import Path
from html.parser import HTMLParser
import base64


class ImgParser(HTMLParser):
    def __init__(self):
        super().__init__()
        self.imagenes = []

    def handle_starttag(self, tag, attrs):
        if tag.lower() == "img":
            atributos = dict(attrs)

            if "src" in atributos:
                self.imagenes.append(atributos["src"])


def obtener_imagenes_html(html_file):
    parser = ImgParser()

    with open(html_file, "r", encoding="utf-8") as f:
        parser.feed(f.read())

    return parser.imagenes


def convertir_base64(imagen_path):
    with open(imagen_path, "rb") as img:
        contenido = img.read()

    return base64.b64encode(contenido).decode("utf-8")


def procesar_html(html_file):

    html_path = Path(html_file)

    imagenes = obtener_imagenes_html(html_path)

    resultado = []

    for img_src in imagenes:

        ruta_imagen = (html_path.parent / img_src).resolve()

        if ruta_imagen.exists():

            base64_img = convertir_base64(ruta_imagen)

            resultado.append({
                "imagen": str(ruta_imagen),
                "base64": base64_img
            })

        else:
            resultado.append({
                "imagen": str(ruta_imagen),
                "error": "Imagen no encontrada"
            })

    return resultado
```
